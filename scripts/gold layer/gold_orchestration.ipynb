```python
import psutil
import os
```

```python
def get_memory_usage():
    process = psutil.Process(os.getpid())
    # Retorna o uso em Megabytes (MB)
    return process.memory_info().rss / (1024 * 1024)
```

```python
notebooks = [
    "./gold_dim_customers",
    "./gold_dim_products",
    "./gold_fact_sales"
]

for nb in notebooks:
    mem_antes = get_memory_usage()
    print(f"Memória inicial: {mem_antes:.2f} MB")
    
    try:
        dbutils.notebook.run(nb, timeout_seconds=0)
        print(f"Notebook {nb} concluído com sucesso.")
    except Exception as e:
        print(f"Erro em {nb}: {e}")
        
    mem_depois = get_memory_usage()
    print(f"Memória após {nb}: {mem_depois:.2f} MB")
    print(f"Consumo incremental: {mem_depois - mem_antes:.2f} MB")
    print("-" * 30)
```
