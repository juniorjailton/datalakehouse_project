#Maping de Volume Files


```python
f = dbutils.fs.ls("/Volumes/workspace/bronze/source_systems")

folders = [
  folder.path for folder in f 
]

all_files = []

for file in folders:
  files = dbutils.fs.ls(file)
  all_files.extend(files)

display(all_files)
```

#Reading from CSV File

```python

dfs = [
  
  (file.path, spark.read.option("header", True)
  .option("inferSchema", True)
  .csv(file.path))
  
  for file in all_files
]


#for file_path, df in dfs:
 #   print(file_path)
 #   display(df)
```

#Writ it to Bronze Layer

```python

for file_path, df in dfs:
    table_name = file_path.split('/')[-1].split(".")[0]
    dtables_names = f"workspace.bronze.{table_name}"
    df.write.mode("overwrite").saveAsTable(dtables_names)

```
